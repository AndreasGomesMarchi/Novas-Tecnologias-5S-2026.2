# Relação entre a variável "sex" e "Survived". 

### Tabela de gráficos

```python
pd.crosstab(df['Sex'], df['Survived'])
```

### Gráfico

```python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Sobrevivência por Sexo')
plt.show()
```

### Sim, as mulheres sobreviveram mais do que os homens
